## Dot Product

```python
# Two example vectors in R^10
a = np.round(rdm.randn(10),3)
b = np.round(rdm.randn(10),3)

# Compute the dot product
dot = sum(a * b)
print(dot)

# Verify with @ function
dot_at = a @ b
print(dot_at)

# Again with numpy
dot_np = np.dot(a,b)
print(dot_np)
```
